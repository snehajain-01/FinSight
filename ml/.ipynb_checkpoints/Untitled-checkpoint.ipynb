```python
import json
import pandas as pd

data = []

for messages in df["messages"]:
    user_message = None
    assistant_message = None

    for message in messages:
        if message["role"] == "user":
            user_message = message["content"]

        elif message["role"] == "assistant":
            assistant_message = message["content"]

    if user_message and assistant_message:
        try:
            # Extract only the actual financial message
            transaction_text = user_message.split(
                "Extract the financial entities from the following message:"
            )[-1].strip()

            # Convert assistant JSON string into dictionary
            result = json.loads(assistant_message)

            data.append({
                "transaction": transaction_text,
                "category": result.get("category")
            })

        except Exception:
            pass

ml_df = pd.DataFrame(data)

print("Shape:", ml_df.shape)
display(ml_df.head(10))
```

Preprocesssing

```python
import re

def clean_text(text):
    text = str(text).lower()
    
    # Remove numbers
    text = re.sub(r'\d+', ' ', text)
    
    # Remove special characters
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)
    
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text

ml_df["clean_transaction"] = ml_df["transaction"].apply(clean_text)

display(
    ml_df[["transaction", "clean_transaction", "category"]].head(10)
)
```

```python
for category in ["food", "shopping", "travel", "bills", "healthcare"]:
    print("\n==========", category.upper(), "==========")
    display(
        ml_df[ml_df["category"] == category][
            ["transaction", "category"]
        ].sample(
            min(5, len(ml_df[ml_df["category"] == category])),
            random_state=42
        )
    )
```

```python
import json
import pandas as pd

data = []

for messages in df["messages"]:

    user_message = None
    assistant_message = None

    for message in messages:
        if message["role"] == "user":
            user_message = message["content"]

        elif message["role"] == "assistant":
            assistant_message = message["content"]

    if user_message and assistant_message:

        try:
            # The actual transaction comes after the blank line
            parts = user_message.split("\n\n", 1)

            if len(parts) == 2:
                transaction_text = parts[1].strip()
            else:
                transaction_text = user_message.strip()

            # Convert assistant response from JSON string to dictionary
            result = json.loads(assistant_message)

            category = result.get("category")

            data.append({
                "transaction": transaction_text,
                "category": category
            })

        except (json.JSONDecodeError, TypeError):
            pass

ml_df = pd.DataFrame(data)

print("Total records:", len(ml_df))
print("\nMissing categories:")
print(ml_df["category"].isna().sum())

display(ml_df.head(10))
```

```python
ml_df = ml_df.dropna(subset=["category"]).copy()

print("Records available for ML:", len(ml_df))
```

```python
import re

def clean_transaction(text):
    text = str(text).lower()

    # Replace account numbers such as XX9806
    text = re.sub(r'\bxx\d+\b', ' ', text)

    # Remove amounts
    text = re.sub(r'rs\.?\s*[\d,]+(?:\.\d+)?', ' ', text)
    text = re.sub(r'inr\s*[\d,]+(?:\.\d+)?', ' ', text)

    # Remove remaining standalone numbers
    text = re.sub(r'\b\d+(?:\.\d+)?\b', ' ', text)

    # Remove punctuation
    text = re.sub(r'[^\w\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()

    return text

ml_df["clean_transaction"] = ml_df["transaction"].apply(clean_transaction)

display(
    ml_df[["transaction", "clean_transaction", "category"]].head(20)
)
```

```python
display(
    ml_df.sample(20, random_state=42)[
        ["transaction", "clean_transaction", "category"]
    ]
)
```

Train Test Split

```python
from sklearn.model_selection import train_test_split

X = ml_df["clean_transaction"]
y = ml_df["category"]

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
#TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=10000,
    ngram_range=(1, 2),
    min_df=2,
    sublinear_tf=True
)

X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)

print("Training shape:", X_train_tfidf.shape)
print("Testing shape:", X_test_tfidf.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model_lr = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

model_lr.fit(X_train_tfidf, y_train)

print("Logistic Regression training completed!")
```

```python
y_pred_lr = model_lr.predict(X_test_tfidf)

print(y_pred_lr[:20])
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(y_test, y_pred_lr)

print("Accuracy:", accuracy)

print("\nClassification Report:")
print(classification_report(y_test, y_pred_lr))
```

```python
from sklearn.svm import LinearSVC

model_svm = LinearSVC(
    class_weight="balanced",
    random_state=42
)

model_svm.fit(X_train_tfidf, y_train)

print("Linear SVM training completed!")
```

```python
y_pred_svm = model_svm.predict(X_test_tfidf)

print(y_pred_svm[:20])
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy_svm = accuracy_score(y_test, y_pred_svm)

print("SVM Accuracy:", accuracy_svm)

print("\nSVM Classification Report:")
print(classification_report(y_test, y_pred_svm))
```
